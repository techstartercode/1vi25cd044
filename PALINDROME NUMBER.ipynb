```python

class Solution:
    def isPalindrome(self, x):
        if x < 0:
            print(False)
            return False

        original = x
        reversed_num = 0

        while x > 0:
            digit = x % 10
            reversed_num = reversed_num * 10 + digit
            x //= 10

        result = original == reversed_num
        print(result)
        return result


# Test the code
obj = Solution()
obj.isPalindrome(121)
obj.isPalindrome(-121)
obj.isPalindrome(10)
```
