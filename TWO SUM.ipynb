```python
class Solution:
    def twoSum(self, nums, target):
        seen = {}

        for i, num in enumerate(nums):
            complement = target - num

            if complement in seen:
                result = [seen[complement], i]
                print(result)
                return result

            seen[num] = i
```

```python
obj = Solution()
obj.twoSum([2, 7, 11, 15], 9)
```
