```python
from typing import List
class Solution:
    def longestCommonPrefix(self, strs: List[str]) -> str:
        prefix = strs[0]

        for s in strs[1:]:
            while not s.startswith(prefix):
                prefix = prefix[:-1]

                if not prefix:
                    return ""

        return prefix
```

```python
obj = Solution()
print(obj.longestCommonPrefix(["flower", "flow", "flight"]))
print(obj.longestCommonPrefix(["dog", "racecar", "car"]))
```
